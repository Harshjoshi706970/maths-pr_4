```python
import pandas as pd 
import numpy as np

df=pd.read_csv('C:/Users/Lenovo/Desktop/Maths/pr 4/student_scores_200.csv')
df
```

1.	Represent each student’s subject scores as a vector.

```python
subjects = ["Math","Science","English","History"]
v=[]
for i in range(200):
  student_vector = df.loc[i,subjects].astype(float).tolist()
  v.append(student_vector)

choice=int(input('enter student index for its vector: '))
print(v[choice])
```

2.	Compute:
○	Norm-1 and Norm-2 of vectors.

```python
choice=int(input('enter index:'))
l1=np.linalg.norm(v[choice],ord=1)
l2=np.linalg.norm(v[choice],ord=2)
print('student vector:',v[choice])
print('norm1:',l1)
print('norm2:',l2)
```

○	Dot product and angle between two students’ score vectors.

```python
s1=int(input(' first student index: '))
s2=int(input(' second student index: '))
dot=np.dot(v[s1],v[s2])
mag_1=np.linalg.norm(v[s1])
mag_2=np.linalg.norm(v[s2])
angle=dot/(mag_1*mag_2)
print('dot product:',dot)
print(np.degrees(np.arccos(angle)))
print(dot)
```

○	Cross product (for 3D selected subjects).

```python
subjects_3d=['Math','Science','English']
choice1=int(input(' first student index: '))
choice2=int(input(' second student index: '))
vec1=df.loc[choice1,subjects_3d].astype(float).values
vec2=df.loc[choice2,subjects_3d].astype(float).values
cross_product=np.cross(vec1,vec2)
print(cross_product)
```

3.	Find the projection of one vector onto another.

```python
c1=int(input('first student index: '))
c2=int(input('second student index: '))

dot=np.dot(v[c1],v[c2])
mag_b=np.linalg.norm(v[c2])
vector_dot=dot/(mag_b**2)
proj=np.dot(vector_dot,v[c2])
mag=np.linalg.norm(proj)
print('the projection of vector1 on to vector2:',proj)
print('the magnitude of the projection:',mag)
```

4.	Form a matrix of students × subjects. Perform:

○	Matrix addition and multiplication.

```python
matrix = np.array(v, dtype=np.float64)
print(matrix[:4])
mat_add=matrix+matrix
print('Matrix addition',mat_add[:4])
mat_mul=matrix @ matrix.T
print('Matrix multiplication',mat_mul[0])
```

○	Transpose and Inverse (if possible).

```python
transpose=matrix.T
print(transpose[0][:20])

# inverse
small_mat=matrix[:4]
determinant=np.linalg.det(small_mat)
# determiant is non-zero so inverse of matrix is possible
inverse=np.linalg.inv(small_mat)
print(inverse)
```

○	Determinant.

```python
small_matrix=matrix[:4] # convertin 200,5 into 5,5
determinant=np.linalg.det(small_matrix)
print(determinant)
```

7.	Compute the eigenvalues and eigenvectors of the covariance matrix.

```python
cov_mat=np.cov(matrix,rowvar=False)
new=np.round(cov_mat,2)
eigenvalues,eigenvectors=np.linalg.eig(new)
print(eigenvalues)
print(eigenvectors)
```
